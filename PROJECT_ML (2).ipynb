```python
from pandas.core import groupby
features={"Sex","RestingECG","ST_Slope","ExerciseAngina","FastingBS","ChestPainType"}
for feature in features:
  print("\n",feature)
  print(df. groupby(feature)["HeartDisease"].mean()*100)
```

```python
# Insights from Analysis:
#  Older patients tend to have a higher rate of heart disease.
# Exerciseangina is associated with a higher rate of heart disease.
# Heart disease in males is more common than in females in this dataset.
#  ASY chest pain has a higher rate of heart disease.
# Flat and Down ST slopes have higher heart disease rates.
# FastingBS = 1 has a higher rate of heart disease.
```

```python
numeric_columns= df.select_dtypes(include=np.number).columns
print(numeric_columns)
for column in numeric_columns :
  Q1=df[column].quantile(0.25)
  Q2=df[column].quantile(0.5)
  Q3=df[column].quantile(0.75)
  IQR=Q3-Q1
  lower_min=Q1-(1.5*IQR)
  upper_max=Q3+(1.5*IQR)
  outliers=df[
      (df[column]<lower_min)|
      (df[column]>upper_max)
  ]

  print("column: ",column,)
  print("lower_min :",lower_min )
  print("upper_max :",upper_max )
  print("number of outliers: ",len(outliers))
```

```python
df['Sex'] = df['Sex'].map({'M': 1, 'F': 0})
df['ExerciseAngina'] = df['ExerciseAngina'].map({'Y': 1, 'N': 0})

# 2. عمل One-Hot Encoding للمتغيرات النصية متعددة الفئات
multi_cat_cols = ['ChestPainType', 'RestingECG', 'ST_Slope']
df = pd.get_dummies(df, columns=multi_cat_cols, drop_first=True)
```

```python
df.head()
```

```python
# بيحول البيانات لنصوص وفي نفس الوقت بيطلع الناتج 0 و 1
df = pd.get_dummies(df, columns=['ChestPainType_ATA'	,'ChestPainType_NAP',	'ChestPainType_TA'	,'RestingECG_Normal'	,'RestingECG_ST'	,'ST_Slope_Flat'	,'ST_Slope_Up'], drop_first=True, dtype=int)
```

```python
df.head()
```

```python
# 1. استبدال قيم 0 في الضغط والكوليسترول بـ Median
df['RestingBP'] = df['RestingBP'].replace(0, df['RestingBP'].median())
df['Cholesterol'] = df['Cholesterol'].replace(0, df['Cholesterol'].median())

# 2. استبدال القيم السلبية في Oldpeak بـ 0 (نفس فكرة replace بشرط)
df.loc[df['Oldpeak'] < 0, 'Oldpeak'] = 0
```

Logistic Rrgression

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score,accuracy_score,recall_score,precision_score,confusion_matrix
from sklearn.preprocessing import StandardScaler
```

```python
x=df.drop("HeartDisease",axis=1)
y=df["HeartDisease"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr_pred=lr.predict(x_test)
acc_lr=accuracy_score(y_test,lr_pred)
print("Accuracy Score:",acc_lr)
```

```python
recall_lr=recall_score(y_test,lr_pred)
print("Recall Score:",recall_lr)
precision_lr=precision_score(y_test,lr_pred)
print("Precision Score:",precision_lr)
f1_lr=f1_score(y_test,lr_pred)
print("F1 Score:",f1_lr)
```

SVM

```python
from sklearn.svm import SVC
```

```python
svm=SVC(
    C=1.0,
    kernel='poly',
    degree=3,
    probability=True

)
svm.fit(x_train_scaled,y_train)
```

```python
svm_pred=svm.predict(x_test_scaled)
acc_svm=accuracy_score(y_test,svm_pred)
print("Accuracy Score:",acc_svm)
```

```python
recall_svm=recall_score(y_test,svm_pred)
print("Recall Score:",recall_svm)
precision_svm=precision_score(y_test,svm_pred)
print("Precision Score:",precision_svm)
```

```python
cm=confusion_matrix(y_test,svm_pred)
print(cm)
```

**Esamble**

```python
from sklearn.ensemble import VotingClassifier
```

```python
classifiers=[
    ("lr",lr),
    ("svm",svm),
    ("dt",dt),
    ("knn",knn),
    ("gnb",gnb)
]
```

```python
vc=VotingClassifier(estimators=classifiers,voting="soft")
vc.fit(x_train_scaled,y_train)
```

```python
vc_pred=vc.predict(x_test_scaled)
acc_vc=accuracy_score(y_test,vc_pred)
print("Accuracy Score:",acc_vc)
```

```python
vc_f1=f1_score(y_test,vc_pred)
print("F1 Score:",vc_f1)
vc_recall=recall_score(y_test,vc_pred)
print("Recall Score:",vc_recall)
vc_precision=precision_score(y_test,vc_pred)
print("Precision Score:",vc_precision)
```
